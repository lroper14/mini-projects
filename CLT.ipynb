```python
import numpy as np
import matplotlib.pyplot as plt
from numpy import random
import matplotlib.animation as animation
```

```python
# Generate 10,000 samples from the exponential distribution
samples = np.random.exponential(size=10000)
```

```python
# Plot the sampled values
plt.hist(samples,bins=50, density=True)
plt.show
```

```python
sample_means = []

for _ in range(10000):
    # Take five random samples from the distribution samples
    random_samples = random.choice(samples, size=5)
    # Calculate the mean of those samples
    mean = np.mean(random_samples)
    # Add those sample means to a list
    sample_means.append(mean)

sample_means = np.asarray(sample_means)
```

```python
plt.hist(sample_means, bins=50, density=True)
plt.show()
```

```python
def get_sample_means(sample_size):

    sample_means = []

    for _ in range(10000):
        # Take five random samples from the distribution samples
        random_samples = random.choice(samples, size=sample_size)
        # Calculate the mean of those samples
        mean = np.mean(random_samples)
        # Add those sample means to a list
        sample_means.append(mean)

    sample_means = np.asarray(sample_means)

    return(sample_means)
```

```python
sample_sizes = [5, 10, 30, 100]

for i in sample_sizes:
    sample_means = get_sample_means(i)
    plt.hist(sample_means, bins=50, density=True)
    plt.show()
```
