```python
plt.plot(frequencies)
plt.ylim(top=1, bottom=0)
plt.show
```

```python
def genetic_drift(pop_size=1000,
                  initial_freq=0.5,
                  gens=100):

    # Store total gene copies in the population
    total_alleles = pop_size * 2
    freq = initial_freq
    freq_list = [freq]

    # Go through every generation and sample from the binomial distribution.
    # Number of successes is the number of A alleles inherited.
    for gen in range(gens-1):
        # Store the number of A alleles from the binomial distribution
        num_A_alleles = np.random.binomial(n = total_alleles, p=freq)
        # Calculate the new allele frequency
        freq = num_A_alleles / total_alleles
        # Store this in a list
        freq_list.append(freq)

    return(freq_list)
```

```python
genetic_drift()
```

```python
# Simulate different population sizes
small_pop = genetic_drift(pop_size=20)
medium_pop = genetic_drift(pop_size=100)
large_pop = genetic_drift(pop_size=1000)
very_large_pop = genetic_drift(pop_size=10000)

print(large_pop)
```

```python
plt.plot(small_pop)
plt.plot(medium_pop)
plt.plot(large_pop)
plt.plot(very_large_pop)
plt.legend(["pop = 20", "= 100", "= 1,000", "= 10,000"])
plt.show()
```
